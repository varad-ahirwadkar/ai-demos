```python
from ogx_client import OgxClient
import rich

# Initialize the OGX client
client = OgxClient(
    base_url=OGX_CONNECTION_URL,
    api_key="fake",
)
```

```python
import inspect
print(inspect.getsource(client.responses.delete))
```

```python
# List available models
models = client.models.list()
rich.print(models)

# Extract LLM model details
llm_model = next(
    m for m in models.data
    if m.custom_metadata.get("model_type") == "llm"
)

model_id = llm_model.id
print(f"LLM Model: {model_id}")
```

```python
# Create a model response.
response = client.responses.create(
    model=model_id,
    input="What is the capital of France?",
)
print(response.output_text)
```

```python
# List all responses
responses_list = client.responses.list()
for resp in responses_list:
    print(resp.id)
```

```python
# Get a model response
response = client.responses.retrieve(responses_list.first_id)
rich.print(response)
```

```python
# List input items.
input_items = client.responses.input_items.list(responses_list.first_id)
for item in input_items:
    rich.print(item)
```

```python
# Delete a response
response = client.responses.delete(responses_list.first_id)
rich.print(response)
```

```python
### Creating OpenAI client because cancel operation is unavailable for OGX Client
from openai import OpenAI

openai_client = OpenAI(base_url=OGX_CONNECTION_URL + "/v1", api_key="fake")


resp = client.responses.create(
    model=model_id,
    input="Write a 1000 word history of computing with extensive detail",
    background=True
)

rich.print("---------Response---------")
rich.print("Status: " + resp.status)
rich.print(resp)

rich.print("---------Cancelling the response---------")
cancelled = client.responses.cancel(resp.id)
rich.print("Status: " + cancelled.status)
rich.print(cancelled)
```
