````python
from ogx_client import OgxClient
import rich
````

````python
# Configuration
OGX_CONNECTION_URL = "http://ogxserver-service.llama.svc.cluster.local:8321"
POLICY_FILE = "data/return-policy.txt"
````

````python
# Initialize OGX client
client = OgxClient(base_url=OGX_CONNECTION_URL)
````

````python
# List the available files
uploaded_files = client.files.list()
````

### File operations with localfs

````python
# Upload return policy document
with open(POLICY_FILE, "rb") as f:
    file_info = client.files.create(
        file=("return-policy.txt", f),
        purpose="assistants",
    )

print(f"Uploaded file: {file_info.id}")
````

````python
# List the files
uploaded_files = client.files.list()

for file_obj in uploaded_files.data:
    print(f"File ID: {file_obj.id} | Name: {file_obj.filename}")
````

````python
# Get file metadata
file_metadata = client.files.retrieve(file_id=file_info.id)

print(f"File Status: {file_metadata.status}")
print(f"Created At: {file_metadata.created_at}")
````

````python
# Retrieve file content
file_content = client.files.content(file_id=file_info.id)

# No need to read or decode; it's already a string!
print("--- File Content ---")
print(file_content)
````

````python
# Delete the file
file_metadata = client.files.delete(file_id=file_info.id)
````

````python
# List the available files
uploaded_files = client.files.list()
rich.print(uploaded_files)
````

### File operations with s3 backend
Update the ogxserver deployment with the following environment variables:
```
- name: ENABLE_S3
  value: "s3"
- name: S3_BUCKET_NAME
  valueFrom:
    secretKeyRef:
      name: s3-creds
      key: AWS_S3_BUCKET
- name: AWS_ACCESS_KEY_ID
  valueFrom:
    secretKeyRef:
      name: s3-creds
      key: AWS_ACCESS_KEY_ID
- name: AWS_DEFAULT_REGION
  valueFrom:
    secretKeyRef:
      name: s3-creds
      key: AWS_DEFAULT_REGION
- name: AWS_SECRET_ACCESS_KEY
  valueFrom:
    secretKeyRef:
      name: s3-creds
      key: AWS_SECRET_ACCESS_KEY
- name: S3_ENDPOINT_URL
  valueFrom:
    secretKeyRef:
      name: s3-creds
      key: AWS_S3_ENDPOINT
```
Redeploy ogxserver to apply the updated configuration.
